```python
# MATERI 12 Di SIMPONI --> NoSQL Key-Value Store and Document Store
# Batch Processing Silver to Gold
import boto3
import pandas as pd

s3 = boto3.client(
    "s3",
    endpoint_url="http://seaweedfs:8333",
    aws_access_key_id="admin",
    aws_secret_access_key="secret"
)

try:
    s3.download_file("silver", "flights/flight_data.parquet", "input.parquet")
    df = pd.read_parquet("input.parquet")

    gold = df.groupby(["airline", "departure_airport"]).agg(
        total_flights=("flight_id", "count"),
        avg_ticket_price=("ticket_price", "mean"),
        min_ticket_price=("ticket_price", "min"),
        max_ticket_price=("ticket_price", "max"),
        total_passengers=("passenger_count", "sum"),
        avg_passengers=("passenger_count", "mean"),
        delayed_flights=("status", lambda x: (x == "Delayed").sum()),
        cancelled_flights=("status", lambda x: (x == "Cancelled").sum())
    ).reset_index()

    gold[["avg_ticket_price", "avg_passengers"]] = gold[
        ["avg_ticket_price", "avg_passengers"]
    ].round(2)

    gold.to_parquet("output.parquet", index=False)
    s3.upload_file("output.parquet", "gold", "flights/flight_aggregated.parquet")

    print(f"Berhasil! Total {len(gold)} baris data Gold.")
    display(gold.head(10))

except Exception as e:
    print(f"Terjadi kesalahan: {e}")
```

```python
# Ingestion from Data lake to MongoDB
import boto3
import pandas as pd
from io import BytesIO
from pymongo import MongoClient

s3 = boto3.client(
    "s3",
    endpoint_url="http://seaweedfs:8333",
    aws_access_key_id="admin",
    aws_secret_access_key="secret"
)

obj = s3.get_object(
    Bucket="gold",
    Key="flights/flight_aggregated.parquet"
)
df = pd.read_parquet(BytesIO(obj["Body"].read()))

client = MongoClient(
    "mongodb://admin:secretsecret@mongodb:27017/?authSource=admin"
)

try:
    collection = client["admin"]["flights"]
    records = df.astype(object).where(pd.notna(df), None).to_dict("records")
    if records:
        collection.insert_many(records)
    print(f"Berhasil memasukkan {len(records)} dokumen ke MongoDB.")
finally:
    client.close()
```
